```python
import pandas as pd
import os

INPUT_FILE = "data/url_benign_malicious_file.csv"
OUTPUT_FILE = "data/processed/clean_urls.csv"

os.makedirs("data/processed", exist_ok=True)

chunk_size = 10000
total_rows = 0
missing_urls = 0
missing_targets = 0

# Process only the two useful columns
reader = pd.read_csv(
    INPUT_FILE,
    usecols=["url", "target"],
    dtype={"url": "string"},
    chunksize=chunk_size
)

first_chunk = True

for chunk in reader:

    total_rows += len(chunk)

    missing_urls += chunk["url"].isna().sum()
    missing_targets += chunk["target"].isna().sum()

    # Preserve all rows at this stage.
    # Do not silently discard missing values.
    chunk.to_csv(
        OUTPUT_FILE,
        mode="w" if first_chunk else "a",
        header=first_chunk,
        index=False
    )

    first_chunk = False

    print(f"Processed {total_rows:,} rows")

print("\nCLEANING SUMMARY")
print("Total rows:", total_rows)
print("Missing URLs:", missing_urls)
print("Missing targets:", missing_targets)
print("Saved to:", OUTPUT_FILE)
```

```python
import pandas as pd

file_path = "data/url_benign_malicious_file.csv"

chunk_size = 10000

# Read column names
columns = pd.read_csv(file_path, nrows=0).columns.tolist()

# Count non-missing values per column
non_missing = pd.Series(0, index=columns, dtype="int64")
total_rows = 0

for chunk in pd.read_csv(file_path, chunksize=chunk_size):
    non_missing += chunk.notna().sum()
    total_rows += len(chunk)

# Calculate missing values and percentages
missing_count = total_rows - non_missing

missing_report = pd.DataFrame({
    "Column": columns,
    "Non_Missing": non_missing.values,
    "Missing": missing_count.values,
    "Missing_Percentage": (
        missing_count.values / total_rows * 100
    )
})

missing_report = missing_report.sort_values(
    "Missing_Percentage",
    ascending=False
)

print("Total rows:", total_rows)
display(missing_report)
```

```python
# Identify completely empty columns
all_nan_columns = missing_report.loc[
    missing_report["Missing_Percentage"] == 100,
    "Column"
].tolist()

# Identify columns with more than 95% missing values
high_nan_columns = missing_report.loc[
    missing_report["Missing_Percentage"] > 95,
    "Column"
].tolist()

print("Completely empty columns:", len(all_nan_columns))
print("Columns with >95% missing:", len(high_nan_columns))

print("\nCompletely empty:")
print(all_nan_columns)

print("\nMore than 95% missing:")
print(high_nan_columns)
```

```python
import pandas as pd
import os

INPUT_FILE = "data/url_benign_malicious_file.csv"
OUTPUT_FILE = "data/processed/reduced_dataset.csv"

os.makedirs("data/processed", exist_ok=True)

# Identify columns with more than 95% missing values
columns_to_remove = missing_report.loc[
    missing_report["Missing_Percentage"] > 95,
    "Column"
].tolist()

# Keep all remaining columns
columns_to_keep = [
    col for col in missing_report["Column"]
    if col not in columns_to_remove
]

print("Columns to remove:", len(columns_to_remove))
print("Columns to retain:", len(columns_to_keep))

# Process the large CSV in chunks
first_chunk = True
total_processed = 0

for chunk in pd.read_csv(
    INPUT_FILE,
    usecols=columns_to_keep,
    chunksize=10000
):
    chunk = chunk[columns_to_keep]

    chunk.to_csv(
        OUTPUT_FILE,
        mode="w" if first_chunk else "a",
        header=first_chunk,
        index=False
    )

    total_processed += len(chunk)
    first_chunk = False

    print(f"Processed: {total_processed:,} rows")

print("\nReduced dataset saved:", OUTPUT_FILE)
```

```python
df_check = pd.read_csv(OUTPUT_FILE, nrows=10)

print("Reduced shape:", df_check.shape)
print("Remaining columns:")
print(df_check.columns.tolist())

print("\nTarget distribution in sample:")
print(df_check["target"].value_counts(dropna=False))

display(df_check.head())
```

```python
import pandas as pd
from collections import Counter

FILE = "data/processed/reduced_dataset.csv"
CHUNK_SIZE = 10000

total_rows = 0
target_counts = Counter()
missing_urls = 0
missing_targets = 0

for chunk in pd.read_csv(FILE, chunksize=CHUNK_SIZE):

    total_rows += len(chunk)

    missing_urls += chunk["url"].isna().sum()
    missing_targets += chunk["target"].isna().sum()

    target_counts.update(
        chunk["target"].astype("string").fillna("<MISSING>").value_counts().to_dict()
    )

print("Total rows:", total_rows)
print("Columns:", chunk.columns.tolist())
print("Missing URLs:", missing_urls)
print("Missing targets:", missing_targets)
print("Target distribution:", target_counts)
```
